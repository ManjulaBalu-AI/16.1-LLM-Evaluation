# Stage2 -  input Quries- Generate

```python
import pandas as pd
import ollama
import time


# ============================================================
# 1. OLLAMA MODEL CONFIGURATION
# ============================================================

MODEL_NAME = "llama3.1"

print("Ollama configuration created successfully!")


# ============================================================
# 2. FILE CONFIGURATION
# ============================================================

INPUT_CSV = "ai_teacher_20_questions.csv"
OUTPUT_CSV = "ai_teacher_results.csv"


# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(INPUT_CSV)

print("Dataset loaded")
print(f"Number of questions: {len(df)}")


# ============================================================
# 4. AI TEACHER INSTRUCTIONS
# ============================================================

TEACHER_INSTRUCTIONS = """
You are an expert AI Teacher.

Your job is to teach students, not just provide answers.

Teaching rules:
- Give only a simple explanation of the concept.
- Use layman terms.
- Answer should be precise and exactly 3 short lines.
- Do not add headings.
- Do not add bullet points.
- Do not add extra information.

Response format:
Explain the concept simply.
"""


# ============================================================
# 5. GENERATE ANSWER WITH RETRY
# ============================================================

def generate_answer(question, max_retries=3):

    prompt = f"""
{TEACHER_INSTRUCTIONS}

Student Question:
{question}
"""

    for attempt in range(max_retries):

        try:

            response = ollama.chat(
                model=MODEL_NAME,
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ]
            )

            answer = response["message"]["content"].strip()

            return answer

        except Exception as e:

            print(f"⚠️ Ollama error: {e}")

            if attempt < max_retries - 1:

                wait_time = 2 ** attempt

                print(
                    f"Retrying in {wait_time} seconds..."
                )

                time.sleep(wait_time)

            else:

                return (
                    "The local AI model is unavailable "
                    "after multiple attempts."
                )


# ============================================================
# 6. GENERATE ANSWERS FOR ALL QUESTIONS
# ============================================================

generated_answers = []

for index, row in df.iterrows():

    question = row["User Query"]

    print("=" * 60)
    print(f"Processing {index + 1}/{len(df)}")
    print(f"Question: {question}")

    answer = generate_answer(question)

    generated_answers.append(answer)

    print(f"Generated Answer:\n{answer}")

    # Small delay between requests
    time.sleep(1)


# ============================================================
# 7. APPEND GENERATED ANSWER TO DATAFRAME
# ============================================================

df["GeneratedAnswer"] = generated_answers


# ============================================================
# 8. SELECT FINAL COLUMNS
# ============================================================

df = df[
    [
        "User Query",
        "Teacher Response",
        "GeneratedAnswer"
    ]
]


# ============================================================
# 9. SAVE RESULT
# ============================================================

df.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 10. DISPLAY RESULT
# ============================================================

print("\n" + "=" * 70)
print("AI TEACHER TEST COMPLETED")
print("=" * 70)

print(df.to_string(index=False))

print("\nResult saved to:")
print(OUTPUT_CSV)
```

```python
import pandas as pd

from deepeval import evaluate
from deepeval.metrics import AnswerRelevancyMetric, GEval
from deepeval.test_case import LLMTestCase, SingleTurnParams
from deepeval.evaluate import AsyncConfig
from deepeval.models import OllamaModel


# ============================================================
# 1. LOAD DATA
# ============================================================

df = pd.read_csv("ai_teacher_results.csv")

print("Dataset loaded")
print(f"Number of questions: {len(df)}")


# ============================================================
# 2. OLLAMA EVALUATION MODEL
# ============================================================

judge_model = OllamaModel(
    model="llama3.1",
    base_url="http://localhost:11434",
    temperature=0
)

print("Ollama Llama 3.1 judge model configured!")


# ============================================================
# 3. CREATE TEST CASES
# ============================================================

test_cases = [
    LLMTestCase(
        input=row["User Query"],
        actual_output=row["GeneratedAnswer"],
        expected_output=row["Teacher Response"]
    )
    for _, row in df.iterrows()
]


# ============================================================
# 4. DEFINE CORRECTNESS METRIC
# ============================================================

correctness = GEval(
    name="Answer Correctness",

    model=judge_model,

    criteria="""
    Evaluate whether the generated answer is factually correct
    compared with the expected teacher answer.

    Give a high score when the important meaning and facts are correct.

    Penalize:
    - incorrect facts
    - contradictory information
    - missing important information

    Minor wording differences should not reduce the score.
    """,

    evaluation_params=[
        SingleTurnParams.ACTUAL_OUTPUT,
        SingleTurnParams.EXPECTED_OUTPUT
    ],

    async_mode=False
)


# ============================================================
# 5. DEFINE RELEVANCY METRIC
# ============================================================

relevancy = AnswerRelevancyMetric(
    model=judge_model,

    include_reason=False,

    async_mode=False
)


# ============================================================
# 6. DEFINE HELPFULNESS METRIC
# ============================================================

helpfulness = GEval(
    name="Helpfulness",

    model=judge_model,

    criteria="""
    Evaluate how helpful the generated answer is for a beginner student.

    A helpful answer should:

    1. Directly answer the student's question.
    2. Be easy to understand.
    3. Provide enough explanation.
    4. Avoid unnecessary information.
    5. Help the student understand the concept.
    """,

    evaluation_params=[
        SingleTurnParams.INPUT,
        SingleTurnParams.ACTUAL_OUTPUT
    ],

    async_mode=False
)


# ============================================================
# 7. RUN DEEPEVAL
# ============================================================

print("\nStarting DeepEval evaluation...")

result = evaluate(
    test_cases=test_cases,

    metrics=[
        correctness,
        relevancy,
        helpfulness
    ],

    async_config=AsyncConfig(run_async=False)
)


# ============================================================
# 8. COMPLETED
# ============================================================

print("\n" + "=" * 70)
print("DEEPEVAL EVALUATION COMPLETED")
print("=" * 70)
```

```python
import pandas as pd

from deepeval import evaluate
from deepeval.metrics import AnswerRelevancyMetric, GEval
from deepeval.test_case import LLMTestCase, SingleTurnParams
from deepeval.evaluate import AsyncConfig
from deepeval.models import OllamaModel


# ============================================================
# 1. LOAD DATA
# ============================================================

df = pd.read_csv("ai_teacher_results.csv")

print("Dataset loaded")
print(f"Number of questions: {len(df)}")


# ============================================================
# 2. OLLAMA MODEL
# ============================================================

judge_model = OllamaModel(
    model="llama3.1",
    base_url="http://localhost:11434",
    temperature=0
)

print("Ollama Llama 3.1 judge model configured successfully!")


# ============================================================
# 3. CREATE TEST CASES
# ============================================================

test_cases = []

for _, row in df.iterrows():

    test_case = LLMTestCase(
        input=row["User Query"],
        actual_output=row["GeneratedAnswer"],
        expected_output=row["Teacher Response"]
    )

    test_cases.append(test_case)


print(f"Created {len(test_cases)} test cases")


# ============================================================
# 4. CORRECTNESS METRIC
# ============================================================

correctness = GEval(
    name="Answer Correctness",

    model=judge_model,

    criteria="""
    Evaluate whether the generated answer is factually correct
    compared with the expected teacher answer.

    Give a high score when the important meaning and facts are correct.

    Penalize:
    - incorrect facts
    - contradictory information
    - missing important information

    Minor wording differences should not reduce the score.
    """,

    evaluation_params=[
        SingleTurnParams.ACTUAL_OUTPUT,
        SingleTurnParams.EXPECTED_OUTPUT
    ],

    async_mode=False
)


# ============================================================
# 5. RELEVANCY METRIC
# ============================================================

relevancy = AnswerRelevancyMetric(
    model=judge_model,

    include_reason=False,

    async_mode=False
)


# ============================================================
# 6. HELPFULNESS METRIC
# ============================================================

helpfulness = GEval(
    name="Helpfulness",

    model=judge_model,

    criteria="""
    Evaluate how helpful the generated answer is for a beginner student.

    A helpful answer should:

    1. Directly answer the student's question.
    2. Be easy to understand.
    3. Provide enough explanation.
    4. Avoid unnecessary information.
    5. Help the student understand the concept.

    Give a high score when the answer is clear,
    simple, relevant, and educational.
    """,

    evaluation_params=[
        SingleTurnParams.INPUT,
        SingleTurnParams.ACTUAL_OUTPUT
    ],

    async_mode=False
)


# ============================================================
# 7. RUN DEEPEVAL
# ============================================================

print("\n" + "=" * 70)
print("STARTING DEEPEVAL EVALUATION")
print("=" * 70)

result = evaluate(
    test_cases=test_cases,

    metrics=[
        correctness,
        relevancy,
        helpfulness
    ],

    async_config=AsyncConfig(run_async=False)
)


# ============================================================
# 8. COMPLETED
# ============================================================

print("\n" + "=" * 70)
print("DEEPEVAL EVALUATION COMPLETED")
print("=" * 70)
```

```python

for i, test_result in enumerate(result.test_results):

    metrics_data = test_result.metrics_data

    df.loc[i, "AnswerCorrectness"] = metrics_data[0].score
    df.loc[i, "AnswerRelevancy"] = metrics_data[1].score
    df.loc[i, "Helpfulness"] = metrics_data[2].score


# ============================================================
# 3. CALCULATE OVERALL ANSWER QUALITY
# ============================================================

df["AnswerQuality"] = (
    df["AnswerCorrectness"]
    + df["AnswerRelevancy"]
    + df["Helpfulness"]
) / 3


# ============================================================
# 4. ROUND SCORES
# ============================================================

metric_columns = [
    "AnswerCorrectness",
    "AnswerRelevancy",
    "Helpfulness",
    "AnswerQuality"
]

df[metric_columns] = df[metric_columns].round(3)


# ============================================================
# 5. AVERAGES
# ============================================================

average_correctness = df["AnswerCorrectness"].mean()
average_relevancy = df["AnswerRelevancy"].mean()
average_helpfulness = df["Helpfulness"].mean()
average_quality = df["AnswerQuality"].mean()


# ============================================================
# 6. QUESTION-WISE RESULTS
# ============================================================

print("\n")
print("=" * 120)
print("                         AI TEACHER ANSWER QUALITY")
print("=" * 120)

display_df = df[
    [
        "User Query",
        "AnswerCorrectness",
        "AnswerRelevancy",
        "Helpfulness",
        "AnswerQuality"
    ]
]


# ============================================================
# 8. SAVE FINAL RESULT
# ============================================================

df.to_csv(
    "ai_teacher_evaluation.csv",
    index=False,
    encoding="utf-8-sig"
)

```
