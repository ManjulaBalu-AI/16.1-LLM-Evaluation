```python
# ============================================================
# 1. IMPORTS
# ============================================================

import pandas as pd
import ollama
import time


# ============================================================
# 2. MODEL CONFIGURATION
# ============================================================

MODEL = "llama3.1"

# Ollama runs locally
# Therefore there is no API cost.

INPUT_PRICE = 0.0
OUTPUT_PRICE = 0.0


# ============================================================
# 3. LOAD QUESTIONS
# ============================================================

df = pd.read_csv("ai_teacher_20_questions.csv")

print(f"Total questions: {len(df)}")


# ============================================================
# 4. AI TEACHER PROMPT
# ============================================================

instructions = """
You are an expert AI Teacher.

Explain the concept using simple layman terms.

Keep the answer precise and easy to understand.

Give the answer in approximately 3 lines.
"""


# ============================================================
# 5. STORAGE
# ============================================================

answers = []

latencies = []

input_tokens = []
output_tokens = []
total_tokens = []

input_costs = []
output_costs = []
total_costs = []


# ============================================================
# 6. GENERATE ANSWERS
# ============================================================

for index, row in df.iterrows():

    question = row["User Query"]

    print("=" * 70)
    print(f"Processing {index + 1}/{len(df)}")
    print(f"Question: {question}")

    try:

        # ----------------------------------------------------
        # START TIMER
        # ----------------------------------------------------

        start_time = time.perf_counter()


        # ----------------------------------------------------
        # OLLAMA API CALL
        # ----------------------------------------------------

        response = ollama.chat(
            model=MODEL,

            messages=[
                {
                    "role": "system",
                    "content": instructions
                },
                {
                    "role": "user",
                    "content": question
                }
            ]
        )


        # ----------------------------------------------------
        # END TIMER
        # ----------------------------------------------------

        end_time = time.perf_counter()


        # ----------------------------------------------------
        # LATENCY
        # ----------------------------------------------------

        latency = end_time - start_time


        # ----------------------------------------------------
        # ANSWER
        # ----------------------------------------------------

        answer = response["message"]["content"].strip()


        # ----------------------------------------------------
        # TOKEN USAGE
        # ----------------------------------------------------

        input_token_count = response.get(
            "prompt_eval_count",
            0
        )

        output_token_count = response.get(
            "eval_count",
            0
        )

        total_token_count = (
            input_token_count +
            output_token_count
        )


        # ----------------------------------------------------
        # COST
        # ----------------------------------------------------
        # Ollama is running locally,
        # therefore API cost = 0.

        input_cost = (
            input_token_count / 1_000_000
        ) * INPUT_PRICE

        output_cost = (
            output_token_count / 1_000_000
        ) * OUTPUT_PRICE

        total_cost = input_cost + output_cost


        # ----------------------------------------------------
        # STORE VALUES
        # ----------------------------------------------------

        answers.append(answer)

        latencies.append(latency)

        input_tokens.append(
            input_token_count
        )

        output_tokens.append(
            output_token_count
        )

        total_tokens.append(
            total_token_count
        )

        input_costs.append(
            input_cost
        )

        output_costs.append(
            output_cost
        )

        total_costs.append(
            total_cost
        )


        # ----------------------------------------------------
        # PRINT RESULTS
        # ----------------------------------------------------

        print(f"Answer: {answer}")
        print(f"Latency: {latency:.2f} seconds")
        print(f"Input tokens: {input_token_count}")
        print(f"Output tokens: {output_token_count}")
        print(f"Total tokens: {total_token_count}")
        print(f"Cost: ${total_cost:.6f}")


    except Exception as e:

        print(f"ERROR: {e}")

        answers.append("ERROR")

        latencies.append(0)

        input_tokens.append(0)
        output_tokens.append(0)
        total_tokens.append(0)

        input_costs.append(0)
        output_costs.append(0)
        total_costs.append(0)


# ============================================================
# 7. ADD RESULTS TO DATAFRAME
# ============================================================

df["GeneratedAnswer"] = answers

df["Latency"] = latencies

df["InputTokens"] = input_tokens

df["OutputTokens"] = output_tokens

df["TotalTokens"] = total_tokens

df["InputCost"] = input_costs

df["OutputCost"] = output_costs

df["TotalCost"] = total_costs


# ============================================================
# 8. SAVE RESULTS
# ============================================================

df.to_csv(
    "ai_teacher_evaluation_Operational_metrics_ollama.csv",
    index=False
)


# ============================================================
# 9. DISPLAY DATAFRAME
# ============================================================

print("\n" + "=" * 70)
print("OLLAMA OPERATIONAL EVALUATION COMPLETED")
print("=" * 70)

df[
    [
        "User Query",
        "GeneratedAnswer",
        "Latency",
        "InputTokens",
        "OutputTokens",
        "TotalTokens",
        "InputCost",
        "OutputCost",
        "TotalCost"
    ]
]
```

```python
# Load the NEW results
df = pd.read_csv("ai_teacher_evaluation_Operational_metrics_ollama.csv")

# CONVERT METRIC COLUMNS TO NUMERIC

metric_columns = [
    "Latency",
    "InputTokens",
    "OutputTokens",
    "TotalTokens",
    "InputCost",
    "OutputCost",
    "TotalCost"
]

for column in metric_columns:

    df[column] = pd.to_numeric(df[column], errors="coerce")

# AVERAGE METRICS

average_latency = df["Latency"].mean()
average_input_tokens = df["InputTokens"].mean()
average_output_tokens = df["OutputTokens"].mean()
average_total_tokens = df["TotalTokens"].mean()
average_input_cost = df["InputCost"].mean()
average_output_cost = df["OutputCost"].mean()
average_total_cost = df["TotalCost"].mean()

# TOTAL METRICS

total_latency = df["Latency"].sum()
total_input_tokens = df["InputTokens"].sum()
total_output_tokens = df["OutputTokens"].sum()
total_tokens = df["TotalTokens"].sum()
total_input_cost = df["InputCost"].sum()
total_output_cost = df["OutputCost"].sum()
total_cost = df["TotalCost"].sum()

# CREATE OPERATIONAL METRICS DATAFRAME

operational_metrics_df = pd.DataFrame({

    "Metric": [
        "Latency (seconds)",
        "Input Tokens",
        "Output Tokens",
        "Total Tokens",
        "Input Cost (USD)",
        "Output Cost (USD)",
        "Total Cost (USD)"
    ],

    "Average": [
        average_latency,
        average_input_tokens,
        average_output_tokens,
        average_total_tokens,
        average_input_cost,
        average_output_cost,
        average_total_cost
    ],

    "Total": [
        total_latency,
        total_input_tokens,
        total_output_tokens,
        total_tokens,
        total_input_cost,
        total_output_cost,
        total_cost
    ]
})


# ROUND VALUES

operational_metrics_df["Average"] = (operational_metrics_df["Average"].round(6))
operational_metrics_df["Total"] = (operational_metrics_df["Total"].round(6))

# DISPLAY DATAFRAME
operational_metrics_df
```

```python
operational_metrics_df.to_csv(
    "ai_teacher_evaluation_Operational_metrics_average.csv",
    index=False,
    encoding="utf-8-sig"
)
```
